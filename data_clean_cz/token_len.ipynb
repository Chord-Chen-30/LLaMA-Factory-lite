```python
import json
import matplotlib.pyplot as plt
import numpy as np
from transformers import AutoTokenizer
from tqdm import tqdm


# 配置
jsonl_file = "./data/toolcall_3more_ocr_code_6200_cz_cleaned_1203.json"
token_path = "Qwen/Qwen3-VL-30B-A3B-Thinking"

bin_width = 50

# 加载tokenizer
tokenizer = AutoTokenizer.from_pretrained(token_path)
```

```python
# 读取并统计
lengths = []
with open(jsonl_file, 'r', encoding='utf-8') as f:

    datas = json.load(f)
    lengths = [len(tokenizer.apply_chat_template(data['messages'], tokenize=True)) for data in tqdm(datas)]
    
#     for data in datas:
#         if 'messages' in data:
#             text = " ".join([msg.get('content', '') for msg in data['messages'] if isinstance(msg, dict)])
#             if text:
#                 tokens = tokenizer.encode(text, add_special_tokens=False)
#                 lengths.append(len(tokens))
            


# 绘图
plt.figure(figsize=(10, 5))
max_len = max(lengths)
bins = np.arange(0, max_len + bin_width, bin_width)
plt.hist(lengths, bins=bins, edgecolor='black', alpha=0.7)
plt.xlabel(f'Token len (bin width={bin_width})')
plt.ylabel('Freq')
plt.title('Token len')
plt.grid(True, alpha=0.3)
plt.show()

# 打印统计信息
print(f"总记录数: {len(lengths)}")
print(f"平均长度: {np.mean(lengths):.2f}")
print(f"最小长度: {min(lengths)}")
print(f"最大长度: {max(lengths)}")
```
