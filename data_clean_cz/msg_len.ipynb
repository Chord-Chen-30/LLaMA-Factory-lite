```python
import json
from tqdm import tqdm
import os
from rich import print
from pprint import pp
import matplotlib.pyplot as plt
```

```python
with open("./data/toolcall_3more_ocr_code_6200_cz_cleaned_1203.json") as f:
    datas = json.load(f)
```

```python
pp(datas[1])
```

```python
# 计算每个messages的长度
lengths = [len(d["messages"]) for d in datas]

# 绘制直方图
plt.hist(lengths, bins=range(0, max(lengths)+10, 1), edgecolor='black')
plt.xlabel('Messages Length')
plt.ylabel('Frequency')
plt.title('Frequency Distribution of Messages Length')
plt.grid(True, alpha=0.3)
plt.show()
```
